```python
import pandas as pd
import numpy as np
data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "Marks": [85, 60, np.nan, 45],
    "Attendance": [92, np.nan, 88, 60]
}
df=pd.DataFrame(data)
print(df)
# Check which cells are missing (True = missing)
print(df.isnull())
# Count missing values per column
print(df.isnull().sum())
# Remove any row that has at least one missing value
print(df.dropna)
# Fill missing values with a fixed number (e.g., 0)
print(df.fillna(0))
#fill missing values with columns average
print(df.fillna(df.mean(numeric_only=True)))
```
