```python
import pandas as pd

data = {
    "Nm": ["Anu", "Ravi", "Meera", "Sam"],
    "Mrks": [85, 60, 90, 45],
    "Att": [92, 70, 88, 60]
}
df=pd.DataFrame(data)
# Rename specific columns
df=df.rename(columns={'Nm':'Name','mrks':'Marks','Att':'Attendance'})
print(df)
# Rename ALL columns at once (must give a name for every column, in order)
df.columns = ['name','marks','Attendance']
print(df)
```
