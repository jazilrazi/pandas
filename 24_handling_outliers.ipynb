```python
import pandas as pd

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam", "Riya", "Kiran", "Dev"],
    "Marks": [85, 60, 90, 45, 70, 75, 950]   # 950 is a typo (should be 95)
}
df = pd.DataFrame(data)

# Step 1: See the problem
print(df["Marks"].mean())      # about 196.43 (way too high!)


# Step 2: Find the "normal range" using the IQR method
Q1 = df["Marks"].quantile(0.25)    # 65.0  (25% of values are below this)
Q3 = df["Marks"].quantile(0.75)    # 87.5  (75% of values are below this)
IQR = Q3 - Q1                      # 22.5  (the "middle spread")

lower = Q1 - 1.5 * IQR             # 31.25
upper = Q3 + 1.5 * IQR             # 121.25

# Step 3: Find the outliers
outliers = df[(df["Marks"] < lower) | (df["Marks"] > upper)]
print(outliers)

# Option 1: REMOVE the outlier rows
df_removed = df[(df["Marks"] >= lower) & (df["Marks"] <= upper)]
print(df_removed["Marks"].mean())   # about 70.83

# Option 2: CAP the values (replace with the limit)
df_capped = df.copy()
df_capped["Marks"] = df_capped["Marks"].clip(lower, upper)
print(df_capped["Marks"].mean())    # about 78.04

# Option 3: REPLACE outliers with the median
df_median = df.copy()
median_value = df["Marks"].median()   # 75
df_median.loc[(df_median["Marks"] < lower) | (df_median["Marks"] > upper), "Marks"] = median_value
print(df_median["Marks"].mean())    # about 71.43
```
