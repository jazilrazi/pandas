```python
import pandas as pd

# Wide format - each month is a separate column
wide = pd.DataFrame({
    "City": ["Delhi", "Mumbai"],
    "Jan": [200, 150],
    "Feb": [220, 170]
})
print(wide)

# melt() - turn WIDE into LONG
long = pd.melt(wide, id_vars="City", var_name="Month", value_name="Sales")
print(long)

# --- stack() and unstack() work with MultiIndex-style data ---
df_multi = long.set_index(["City", "Month"])
print(df_multi)

# unstack() - turn the inner index level into COLUMNS (long -> wide)
back_to_wide = df_multi.unstack()
print(back_to_wide)

# stack() - the opposite: turn COLUMNS into an inner index level (wide -> long)
back_to_long = back_to_wide.stack()
print(back_to_long)
```
