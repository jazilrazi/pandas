```python
import pandas as pd

jan_marks = pd.DataFrame({
    "Name": ["Anu", "Ravi"],
    "Marks": [85, 60]
})

feb_marks = pd.DataFrame({
    "Name": ["Meera", "Sam"],
    "Marks": [90, 45]
})

# Stack rows on top of each other (default: axis=0)
combined = pd.concat([jan_marks, feb_marks])
print(combined)

# Fix the messy repeated index (0,1,0,1) after stacking
combined = pd.concat([jan_marks, feb_marks], ignore_index=True)
print(combined)

# Stack side by side instead (axis=1) - adds as new columns
science_marks = pd.DataFrame({"Science": [70, 65, 80, 50]})
side_by_side = pd.concat([combined, science_marks], axis=1)
print(side_by_side)
```
