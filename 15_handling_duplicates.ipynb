```python
import pandas as pd

data = {
    "Name": ["Anu", "Ravi", "Meera", "Ravi"],
    "Marks": [85, 60, 90, 60]
}
df = pd.DataFrame(data)
print(df)

# Check which rows are duplicates (True = it's a repeat of an earlier row)
print(df.duplicated())

# Count how many duplicate rows exist
print(df.duplicated().sum())

# Remove duplicate rows (keeps the first occurrence by default)
df_clean = df.drop_duplicates()
print(df_clean)

# Check duplicates based on ONLY specific columns (e.g., just Name)
print(df.duplicated(subset=["Name"]))

# Remove duplicates but keep the LAST occurrence instead of the first
df_clean2 = df.drop_duplicates(keep="last")
print(df_clean2)
```
