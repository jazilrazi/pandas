Looking at data: head(), tail(), shape, info(), describe()

```python
import pandas as pd
data = {
    "Name": ["Jazil", "Rahul", "Anu", "Meera", "Arun"],
    "Age": [21, 22, 20, 21, 23],
    "Marks": [85, 72, 91, 68, 78]
}
df = pd.DataFrame(data)
print(df.head(1))       #first 1 row
print(df.tail(1))       #last 1 row
print(df.shape)         #(rows,columns)
print(df.info())        #column names,datatype,missing values
print(df.describe())    #statistics:mean,max,min,etc.(numeric columns only)
print(df.to_string())

```
