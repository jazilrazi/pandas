```python
import pandas as pd

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "Marks": [85, 60, 90, 45],
    "Attendance": [92, 70, 88, 60]
}
df = pd.DataFrame(data)

# On a single column
print(df["Marks"].sum())     # total of all marks
print(df["Marks"].mean())    # average marks
print(df["Marks"].max())     # highest mark
print(df["Marks"].min())     # lowest mark
print(df["Marks"].count())   # how many values (non-missing)

# On the whole DataFrame at once (applies to all numeric columns)
print(df.sum(numeric_only=True))
print(df.mean(numeric_only=True))

# Multiple stats together using describe() (recap from Topic 5)
print(df["Marks"].describe())
```
