```python
import pandas as pd

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "Marks": [85, 60, 90, 45]
}
df = pd.DataFrame(data)

# --- lambda basics first ---
# A lambda is just a tiny, one-line function without a name
square = lambda x: x * x
print(square(5))   # 25

# --- Using apply() with a lambda on a column ---
df["Marks_Doubled"] = df["Marks"].apply(lambda x: x * 2)
print(df)

# --- Using apply() with a proper custom function (for more complex logic) ---
def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "F"

df["Grade"] = df["Marks"].apply(get_grade)
print(df)

# --- map() - similar to apply(), but only works on a single Series, often for simple lookups ---
grade_points = {"A": 4, "B": 3, "C": 2, "F": 0}
df["GradePoint"] = df["Grade"].map(grade_points)
print(df)
```
