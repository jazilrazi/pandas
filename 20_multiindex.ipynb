```python
import pandas as pd

data = {
    "City": ["Delhi", "Delhi", "Mumbai", "Mumbai"],
    "Month": ["Jan", "Feb", "Jan", "Feb"],
    "Sales": [200, 220, 150, 170]
}
df = pd.DataFrame(data)

# Set two columns as a MultiIndex
df_multi = df.set_index(["City", "Month"])
print(df_multi)

# Access all rows for one City (outer level)
print(df_multi.loc["Delhi"])

# Access a specific City + Month combination
print(df_multi.loc[("Delhi", "Jan")])

# GroupBy naturally creates a MultiIndex too, when grouping by 2+ columns
grouped = df.groupby(["City", "Month"])["Sales"].sum()
print(grouped)
print(grouped.index)   # shows it's a MultiIndex

# Convert a MultiIndex back to normal flat columns
flat = grouped.reset_index()
print(flat)
```
