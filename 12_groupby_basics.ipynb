```python
import pandas as pd

data = {
    "City": ["Delhi", "Mumbai", "Delhi", "Mumbai", "Chennai"],
    "Product": ["Shoes", "Shoes", "Bags", "Bags", "Shoes"],
    "Sales": [200, 150, 300, 100, 250]
}
df = pd.DataFrame(data)

# Total sales per city
print(df.groupby("City")["Sales"].sum())

# Average sales per city
print(df.groupby("City")["Sales"].mean())

# Count how many rows (orders) per city
print(df.groupby("City")["Sales"].count())

# Group by TWO columns (City + Product)
print(df.groupby(["City", "Product"])["Sales"].sum())
```
