```python
import pandas as pd

data = {
    "City": ["Delhi", "Mumbai", "Delhi", "Mumbai", "Chennai", "Delhi"],
    "Product": ["Shoes", "Shoes", "Bags", "Bags", "Shoes", "Shoes"],
    "Sales": [200, 150, 300, 100, 250, 180]
}
df = pd.DataFrame(data)


# Multiple aggregations at once, on one column
print(df.groupby("City")["Sales"].agg(["sum", "mean", "max", "min"]))

# Different aggregations for different columns
print(df.groupby("City").agg({
    "Sales": "sum",
    "Product": "count"
}))

# Custom function inside agg()
def sales_range(x):
    return x.max() - x.min()

print(df.groupby("City")["Sales"].agg(["sum", sales_range]))

# Naming your result columns clearly
print(df.groupby("City")["Sales"].agg(
    Total_Sales="sum",
    Average_Sales="mean"
))
```
