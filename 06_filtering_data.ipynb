```python
import pandas as pd
data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "Marks": [85, 60, 90, 45],
    "Attendance": [92, 70, 88, 60]
}
df = pd.DataFrame(data)

#students who scored above 75
print (df[df['Marks']>75])

#students with attendance below 75
print (df[df['Marks']<75])

#combining two condition (AND)
# note brackets around each condition
print(df[(df['Marks']>50)&(df['Attendance']>65)])

#combining tow condition (OR)
print(df[(df['Marks']>80) | (df['Attendance']>80)])
```
