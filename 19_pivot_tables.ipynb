```python
import pandas as pd

data = {
    "City": ["Delhi", "Mumbai", "Delhi", "Mumbai", "Chennai", "Delhi"],
    "Product": ["Shoes", "Shoes", "Bags", "Bags", "Shoes", "Shoes"],
    "Sales": [200, 150, 300, 100, 250, 180]
}
df = pd.DataFrame(data)

#Basic pivot table: rows=City, columns=Product, values=Sales (default: mean)
print(pd.pivot_table(df, index="City", columns="Product", values="Sales"))

#Same, but sum instead of average
print(pd.pivot_table(df, index="City", columns="Product", values="Sales", aggfunc="sum"))

# Fill empty combinations (no data) with 0 instead of NaN
print(pd.pivot_table(df, index="City", columns="Product", values="Sales",
                      aggfunc="sum", fill_value=0))
#
# Add row/column totals
print(pd.pivot_table(df, index="City", columns="Product", values="Sales",
                      aggfunc="sum", fill_value=0, margins=True))
```
