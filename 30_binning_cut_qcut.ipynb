```python
import pandas as pd

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam", "Riya", "Kiran"],
    "Marks": [85, 60, 45, 90, 30, 72]
}
df = pd.DataFrame(data)

# pd.cut() - YOU define the exact bin edges (ranges)
df["Band"] = pd.cut(df["Marks"],
                     bins=[0, 40, 70, 100],
                     labels=["Low", "Medium", "High"])
print(df)


# pd.qcut() - pandas splits data into EQUAL-SIZED groups automatically (by quantile)
df["Quartile_Band"] = pd.qcut(df["Marks"], q=3, labels=["Bottom", "Middle", "Top"])
print(df)

# cut() without labels - shows the actual ranges
print(pd.cut(df["Marks"], bins=[0, 40, 70, 100]))
```
