```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "Marks": [85, 60, 90, 45]
}
df = pd.DataFrame(data)
# Using a NumPy function directly on a pandas column
df["Sqrt_Marks"] = np.sqrt(df["Marks"])
print(df)

# np.where() - like an if/else, but works on the whole column at once (faster than apply())
df["Result"] = np.where(df["Marks"] >= 50, "Pass", "Fail")
print(df)
# np.select() - for MULTIPLE conditions (like if/elif/elif/else)
conditions = [
    df["Marks"] >= 90,
    df["Marks"] >= 75,
    df["Marks"] >= 50
]
choices = ["A", "B", "C"]
df["Grade"] = np.select(conditions, choices, default="F")
print(df)
# Using np.nan to represent missing data (same NaN pandas uses)
df.loc[1, "Marks"] = np.nan
print(df)
# Rounding numbers using NumPy
df["Sqrt_Marks_Rounded"] = np.round(df["Sqrt_Marks"], 2)
print(df)
```
