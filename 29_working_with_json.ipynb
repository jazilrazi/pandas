```python
import pandas as pd

# A JSON-style Python list of dictionaries (like what an API might return)
json_data = [
    {"Name": "Anu", "Marks": 85, "City": "Delhi"},
    {"Name": "Ravi", "Marks": 60, "City": "Mumbai"},
    {"Name": "Meera", "Marks": 90, "City": "Chennai"}
]


# Convert directly into a DataFrame
df = pd.DataFrame(json_data)
print(df)

# Reading a JSON FILE directly
df_from_file = pd.read_json("students.json")
print(df_from_file)

# Saving a DataFrame back to JSON
df.to_json("students_output.json", orient="records")

# Handling NESTED JSON (a dictionary inside a dictionary)
nested_json = [
    {"Name": "Anu", "Address": {"City": "Delhi", "Pin": 110001}},
    {"Name": "Ravi", "Address": {"City": "Mumbai", "Pin": 400001}}
]
df_nested = pd.json_normalize(nested_json)
print(df_nested)
```
