```python
import pandas as pd
data={
     'name':['anu','ravi','meera','sam'],
     'marks':[85,60,90,45],
     'attendence':[92,70,88,60]
 }
df = pd.DataFrame(data)

 #selecting a column
print(df['marks'])

 #selecting multiple columns
print(df[['name','marks']])

 # selecting a row by LABEL (loc)
print(df.loc[1])         #row with index label 1 (ravi)

#selecting a row by POSITION (iloc)
print(df.iloc[1])         #row at position 1 (also ravi,same here)

#selecting specific rows + column together
print(df.loc[0:2,['name','marks']])   #rows 0 to 2, only name & marks
print(df.to_string())
```
