```python
import pandas as pd
import matplotlib.pyplot as plt   # pandas uses this under the hood

data = {
    "City": ["Delhi", "Mumbai", "Chennai", "Kolkata"],
    "Sales": [500, 350, 250, 400]
}
df = pd.DataFrame(data)

# Bar chart - great for comparing categories
df.plot(x="City", y="Sales", kind="bar", title="Sales by City")
plt.show()

# Line chart - great for trends over time
daily_sales = pd.Series([100, 120, 90, 150, 130],
                         index=["Mon", "Tue", "Wed", "Thu", "Fri"])
daily_sales.plot(kind="line", title="Daily Sales Trend")
plt.show()

# Histogram - shows how values are distributed
marks = pd.Series([85, 60, 90, 45, 70, 75, 95, 50, 65, 80])
marks.plot(kind="hist", title="Marks Distribution")
plt.show()

# Pie chart - shows proportion of a whole
df.plot(x="City", y="Sales", kind="pie", title="Sales Share")
plt.show()
```
