```python
import pandas as pd

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "Marks": [85, 60, 90, 45],
    "Attendance": [92, 70, 88, 60]
}
df = pd.DataFrame(data)

# Add a new column with a fixed calculation
df["Bonus"] = 5
df["Total"] = df["Marks"] + df["Bonus"]
print(df)

# Add a column based on a condition
df["Result"] = ["Pass" if m >= 50 else "Fail" for m in df["Marks"]]
print(df)

# Delete a column
df = df.drop("Bonus", axis=1)
print(df)

# Delete multiple columns
df = df.drop(["Bonus", "Attendance"], axis=1, errors="ignore")
print(df)
```
