```python
import pandas as pd

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "Marks": ["85", "60", "90", "45"],   # notice: these are TEXT, not numbers
    "Attendance": [92, 70, 88, 60]
}
df = pd.DataFrame(data)

# Check current data types
print(df.dtypes)

# Convert Marks from text to whole numbers
df["Marks"] = df["Marks"].astype(int)
print(df.dtypes)

# Convert to decimal numbers instead
df["Attendance"] = df["Attendance"].astype(float)
print(df.dtypes)

# Convert a number column to text (sometimes needed for merging with text IDs)
df["Name_ID"] = df["Marks"].astype(str) + "_" + df["Name"]
print(df)
```
