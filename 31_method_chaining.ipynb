```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam", "Riya"],
    "Marks": [85, 60, np.nan, 45, 92],
    "Attendance": [92, 70, 88, 60, 95]
}
df = pd.DataFrame(data)


# --- WITHOUT chaining (step by step, lots of variables) ---
df1 = df.dropna()
df2 = df1[df1["Marks"] >= 50]
df3 = df2.sort_values("Marks", ascending=False)
result = df3[["Name", "Marks"]]
print(result)

# --- WITH chaining (same result, one connected flow) ---
result_chained = (
    df.dropna()
      .query("Marks >= 50")
      .sort_values("Marks", ascending=False)
      [["Name", "Marks"]]
)
print(result_chained)
```
