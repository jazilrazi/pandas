```python
import pandas as pd

data = {
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "TestDate": ["2024-01-15", "2024-02-20", "2024-01-28", "2024-03-05"]
}
df = pd.DataFrame(data)

# Check current data type (it's text/object right now)
print(df.dtypes)

# Convert text to proper datetime
df["TestDate"] = pd.to_datetime(df["TestDate"])
print(df.dtypes)

# Extract parts of the date
df["Year"] = df["TestDate"].dt.year
df["Month"] = df["TestDate"].dt.month
df["Day"] = df["TestDate"].dt.day
df["DayName"] = df["TestDate"].dt.day_name()  # e.g., "Monday"
print(df)

# Filter using dates - tests only in January
print(df[df["TestDate"].dt.month == 1])

# Calculate days between a date and today
today = pd.Timestamp("2024-03-10")
df["DaysAgo"] = (today - df["TestDate"]).dt.days
print(df)
```
