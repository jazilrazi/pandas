```python
import pandas as pd

marks = pd.DataFrame({
    "StudentID": [1, 2, 3, 4],
    "Name": ["Anu", "Ravi", "Meera", "Sam"],
    "Marks": [85, 60, 90, 45]
})

attendance = pd.DataFrame({
    "StudentID": [1, 2, 3, 5],
    "Attendance": [92, 70, 88, 99]
})

# Inner join - only students present in BOTH tables
print(pd.merge(marks, attendance, on="StudentID", how="inner"))

# Left join - keep ALL students from marks, fill missing attendance with NaN
print(pd.merge(marks, attendance, on="StudentID", how="left"))

# Right join - keep ALL students from attendance
print(pd.merge(marks, attendance, on="StudentID", how="right"))

# Outer join - keep EVERY student from both tables
print(pd.merge(marks, attendance, on="StudentID", how="outer"))
```
