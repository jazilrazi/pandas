```python
import pandas as pd

data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Sales": [100, 120, 90, 150, 130, 170, 160]
}
df = pd.DataFrame(data)


# Rolling: 3-day moving average
df["Rolling_Avg_3"] = df["Sales"].rolling(window=3).mean()

# Rolling: 3-day moving sum
df["Rolling_Sum_3"] = df["Sales"].rolling(window=3).sum()

# Expanding: running average from the start up to each row
df["Expanding_Avg"] = df["Sales"].expanding().mean()

# Expanding: running total (cumulative sum)
df["Running_Total"] = df["Sales"].expanding().sum()

print(df)
```
