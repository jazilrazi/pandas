```python
import pandas as pd
import numpy as np
import time

# Create a large dataset for testing
df = pd.DataFrame({"Marks": np.random.randint(0, 100, size=100000)})


# --- SLOW WAY: using a loop ---
start = time.time()
bonus_marks = []
for m in df["Marks"]:
    bonus_marks.append(m + 5)
df["Bonus_Loop"] = bonus_marks
print("Loop time:", time.time() - start)

# --- FAST WAY: vectorized operation ---
start = time.time()
df["Bonus_Vectorized"] = df["Marks"] + 5
print("Vectorized time:", time.time() - start)

# --- Also slow: apply() (better than a loop, but still row-by-row) ---
start = time.time()
df["Bonus_Apply"] = df["Marks"].apply(lambda x: x + 5)
print("Apply time:", time.time() - start)
```
