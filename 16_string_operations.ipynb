```python
import pandas as pd

data = {
    "Name": [" ANU", "ravi ", "Meera", "SAM"],
    "City": ["Delhi", "mumbai", "CHENNAI", "delhi"]
}
df = pd.DataFrame(data)

# Convert to lowercase
print(df["Name"].str.lower())

# Convert to uppercase
print(df["Name"].str.upper())

# Remove extra spaces from both ends
print(df["Name"].str.strip())

# Check which names contain a specific letter/word (case-sensitive by default)
print(df["Name"].str.contains("a", case=False))

# Replace part of a string
print(df["City"].str.replace("delhi", "Delhi", case=False)

# Combine cleanup steps together
df["Name"] = df["Name"].str.strip().str.title()
print(df)
```
