1. Create a Series from a list of 5 student marks.

```python
import pandas as pd
student_marks = pd.Series(['90','93','92','96','100'])
print(student_marks)
```

2. Create a Series from a dictionary of items and prices.

```python
import pandas as pd
data = {
    'item':['manga','thenga','chakka'],
    'price':['50','60','70']

}
df = pd.Series(data)
print(df)
```
